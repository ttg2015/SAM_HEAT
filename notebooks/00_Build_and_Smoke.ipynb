Run this first on Torch. It checks four things:
- the executable and the files it needs exist;
- the generated prm is the paper's;
- a 2-hour run finishes without NaN;
- **the sun is perpetual**: SOLIN stays constant at 565·cos 42.332° ≈ 418 W/m². Our old runs had a diurnal cycle here.

```python
from samheat.paths import SITE
for name, p in [("executable", SITE.build / SITE.executable), ("RUNDATA", SITE.rundata_source / "RUNDATA"),
                ("stat2nc", SITE.util_dir / "stat2nc"), ("work_root", SITE.work_root.parent)]:
    print(f"{name:11s} {'OK ' if p.exists() else 'MISSING'} {p}")
import sys; print("python for AUTO_ADVANCE:", sys.executable)
```
